```python
import sqlite3
```

```python
nome_banco = 'cadastro.db'
```

```python
#INSERIR DADOS NA TABELA
script_cadastrar = "INSERIR INTO (username, email, password) VALUS )?,?,?"

try:
    with sqlite3.connect(nome_banco) as conn:

        # Cria um cursor
        cur = conn.cursor()

        # Executa o script
        cur.execute(script_cadastrar, ('Johnny', 'Genial@email.com', 'senhaForte123'))

        # Salva as alterações no banco de dados
        conn.commit()



except sqlite3.OperationalError as e:
    print("ERRO: ", e)
```

```python
script_consulta = "SELECT * FROM usuarios"

try:
    with sqlite3.connect(nome_banco) as conn:
        # Cria um cursor
        cur = conn.cursor()
        
        
        # Executa o script
        cur.execute(script_consulta)
        res = cur.fetchall() # retorna uma lista de listas
        
        print(res)
except sqlite3.OperationalError as e:
    print("ERRO: ", e)
```
